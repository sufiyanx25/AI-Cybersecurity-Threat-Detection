# 04 - Evaluation & Hyperparameter Tuning (Doc Step 7)
Demonstrates ROC analysis, classification reports, and RandomizedSearchCV tuning.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import json, joblib
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
print('Imports OK')
```

```python
meta_path = '../models/metadata.json'
if os.path.exists(meta_path):
    with open(meta_path) as f:
        meta = json.load(f)
    print('Best Hyperparameters (RandomizedSearchCV):', meta['best_hyperparameters_rf'])
    print('\nAll Model Metrics:')
    for model_name, metrics in meta['metrics'].items():
        print(f'  {model_name}: {metrics}')
else:
    print('Run run_all.py first to generate metadata.json')
```

```python
roc_path = '../outputs/reports/roc_curves.json'
if os.path.exists(roc_path):
    with open(roc_path) as f:
        roc_data = json.load(f)
    fig, ax = plt.subplots(figsize=(7, 5))
    rf = roc_data.get('deploy_rf', {})
    lr = roc_data.get('logistic_regression', {})
    if rf:
        ax.plot(rf['fpr'], rf['tpr'], label=f"Deploy RF AUC={rf['auc']:.3f}")
    if lr:
        ax.plot(lr['fpr'], lr['tpr'], linestyle='--', label=f"LR AUC={lr['auc']:.3f}")
    ax.plot([0,1],[0,1], color='grey', linestyle=':')
    ax.set_title('ROC Curves'); ax.legend()
    plt.tight_layout()
    plt.savefig('../outputs/plots/nb_roc.png', dpi=150)
    plt.show()
    print('ROC AUC (Deployment RF):', rf.get('auc'))
else:
    print('Run run_all.py first to generate ROC data')
```
