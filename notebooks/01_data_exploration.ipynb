## AI-Powered Cybersecurity Threat Detection System
Explores the synthetic NSL-KDD/CICIDS-style network traffic dataset generated for this project.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns
from src.data.generate_dataset import generate_network_traffic
print('Imports OK')
```

```python
df = generate_network_traffic(num_rows=2000, random_seed=42)
print('Shape:', df.shape)
df.head(5)
```

```python
print('Attack Label Distribution:')
print(df['Attack_Label'].value_counts())
print('\nMissing Values:')
print(df.isnull().sum())
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
df['Attack_Label'].value_counts().plot(kind='bar', ax=ax, color='steelblue')
ax.set_title('Traffic Class Distribution')
ax.set_xlabel('Attack Label')
ax.set_ylabel('Count')
plt.tight_layout()
plt.savefig('../outputs/plots/nb_class_distribution.png', dpi=150)
plt.show()
print('Plot saved.')
```

```python
print('\nDoc 3-feature statistics per class:')
df.groupby('Attack_Label')[['packet_size','failed_logins','request_frequency']].mean().round(2)
```
